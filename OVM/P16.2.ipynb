```python
import numpy as np
from math import comb

# Problem parameters
S0 = 3
E = 2
T = 1
r = 0.05
sigma = 0.3
M = 400

# Time step and probability
dt = T / M
p = 0.5

# Upward and downward multipliers
u = np.exp(sigma * np.sqrt(dt) + (r - 0.5 * sigma**2) * dt)
d = np.exp(-sigma * np.sqrt(dt) + (r - 0.5 * sigma**2) * dt)
```

```python
k = np.arange(M + 1)

ST = S0 * u**k * d**(M - k)
payoff = np.maximum(ST - E, 0)
```

```python
prob = np.array([
    comb(M, int(j)) * p**j * (1-p)**(M-j)
    for j in k
])
```

```python
print(prob.sum())
```

```python
option_value = np.exp(-r * T) * np.sum(prob * payoff)

print("Option value:", option_value)
```
